## Preprocess Data

### Subtask:
Separate features (X) and target (y), encode the target labels (prognosis) as needed, and split the data into training and testing sets to prepare it for machine learning models.

**Reasoning**:
The preprocessing steps involve:
1.  **Separating Features and Target**: The `prognosis` column is the target variable, and all other columns are features. This separation is crucial for machine learning model training.
2.  **Encoding Target Labels**: Machine learning models typically require numerical input. The `prognosis` column contains categorical disease names, which need to be converted into numerical labels using `LabelEncoder`.
3.  **Splitting Data**: To evaluate the model's performance on unseen data, the dataset must be split into training and testing sets. `train_test_split` with `test_size=0.3`, `random_state=42`, and `stratify=y` ensures a reproducible and balanced split across disease classes.

**Reasoning**:
To implement the preprocessing steps outlined in the subtask, I will use a single code block to separate features and target, encode the target variable, and split the data into training and testing sets.



```python
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split

# 1. Separate features (X) and target (y)
X = train_df.drop('prognosis', axis=1)
y = train_df['prognosis']

print(f"Shape of X: {X.shape}")
print(f"Shape of y: {y.shape}")

# 2. Encode target labels (prognosis)
label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)

print("\nOriginal target labels (first 5):", y.head().tolist())
print("Encoded target labels (first 5):", y_encoded[:5].tolist())
print("Number of unique encoded labels:", len(label_encoder.classes_))

# 3. Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y_encoded, test_size=0.3, random_state=42, stratify=y_encoded)

print(f"\nShape of X_train: {X_train.shape}")
print(f"Shape of X_test: {X_test.shape}")
print(f"Shape of y_train: {y_train.shape}")
print(f"Shape of y_test: {y_test.shape}")
```

## Train and Evaluate Models

### Subtask:
Train Decision Tree, Random Forest, and Naive Bayes models on the preprocessed training data. For each model, evaluate its accuracy on the test data and print a classification report.


**Reasoning**:
I need to train and evaluate Decision Tree, Random Forest, and Naive Bayes models. This involves importing the necessary libraries, initializing each model, training it on the `X_train` and `y_train` data, making predictions on `X_test`, and then printing the accuracy score and classification report for each model.



```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, classification_report

# 3. Decision Tree
print("\n--- Decision Tree Model ---")
dtc = DecisionTreeClassifier(random_state=42)
dtc.fit(X_train, y_train)
y_pred_dtc = dtc.predict(X_test)
print(f"Accuracy (Decision Tree): {accuracy_score(y_test, y_pred_dtc):.4f}")
print("Classification Report (Decision Tree):\n", classification_report(y_test, y_pred_dtc, zero_division=0))

# 4. Random Forest
print("\n--- Random Forest Model ---")
rfc = RandomForestClassifier(random_state=42)
rfc.fit(X_train, y_train)
y_pred_rfc = rfc.predict(X_test)
print(f"Accuracy (Random Forest): {accuracy_score(y_test, y_pred_rfc):.4f}")
print("Classification Report (Random Forest):\n", classification_report(y_test, y_pred_rfc, zero_division=0))

# 5. Naive Bayes
print("\n--- Naive Bayes Model ---")
gnb = GaussianNB()
gnb.fit(X_train, y_train)
y_pred_gnb = gnb.predict(X_test)
print(f"Accuracy (Naive Bayes): {accuracy_score(y_test, y_pred_gnb):.4f}")
print("Classification Report (Naive Bayes):\n", classification_report(y_test, y_pred_gnb, zero_division=0))
```

## Develop Prediction Function

### Subtask:
Create a Python function that takes a list of symptom names as input, converts these symptoms into a 132-length binary feature vector, and uses the selected best model to predict the disease.


**Reasoning**:
To fulfill the subtask, I need to define a Python function that converts a list of symptoms into a binary feature vector, uses the `best_model` to predict the disease, and then decodes the prediction back to the disease name using the `label_encoder`. This requires `numpy` for array manipulation and access to `X.columns` for mapping symptom names to feature vector indices.



```python
import numpy as np

def predict_disease(symptoms_list):
    # Create a zero-initialized NumPy array of length 132 (number of features in X)
    feature_vector = np.zeros(len(X.columns))

    # Iterate through the symptoms_list and set corresponding feature to 1
    for symptom in symptoms_list:
        if symptom in X.columns:
            feature_vector[X.columns.get_loc(symptom)] = 1
        else:
            print(f"Warning: Symptom '{symptom}' not found in the feature list.")

    # Reshape the feature vector to be a 2D array for prediction
    feature_vector = feature_vector.reshape(1, -1)

    # Use the best_model to predict the encoded disease label
    predicted_encoded_label = best_model.predict(feature_vector)

    # Convert the predicted encoded label back into the original disease name
    predicted_disease = label_encoder.inverse_transform(predicted_encoded_label)

    return predicted_disease[0]

print("Prediction function `predict_disease` defined.")
```

**Reasoning**:
The prediction function `predict_disease` has been successfully defined. To complete the subtask, it's necessary to demonstrate its usage with a sample prediction to ensure it works as expected.



```python
sample_symptoms = ['itching', 'skin_rash', 'nodal_skin_eruptions', 'dischromic _patches']
predicted_disease = predict_disease(sample_symptoms)
print(f"Predicted Disease for symptoms {sample_symptoms}: {predicted_disease}")

sample_symptoms_2 = ['vomiting', 'abdominal_pain', 'diarrhoea']
predicted_disease_2 = predict_disease(sample_symptoms_2)
print(f"Predicted Disease for symptoms {sample_symptoms_2}: {predicted_disease_2}")
```

The previous code generated a UserWarning because the prediction function's input (a numpy array) lacked feature names, which the Random Forest model was fitted with. To resolve this, I will modify the `predict_disease` function to convert the feature vector into a pandas DataFrame with the correct column names before prediction, ensuring consistency with the training data.



```python
import numpy as np
import pandas as pd # Import pandas for DataFrame creation

def predict_disease(symptoms_list):
    # Create a zero-initialized NumPy array of length 132 (number of features in X)
    feature_vector = np.zeros(len(X.columns))

    # Iterate through the symptoms_list and set corresponding feature to 1
    for symptom in symptoms_list:
        if symptom in X.columns:
            feature_vector[X.columns.get_loc(symptom)] = 1
        else:
            print(f"Warning: Symptom '{symptom}' not found in the feature list.")

    # Convert the feature vector to a DataFrame with column names for prediction
    # This ensures the model receives input with feature names, suppressing the warning
    input_df = pd.DataFrame([feature_vector], columns=X.columns)

    # Use the best_model to predict the encoded disease label
    predicted_encoded_label = best_model.predict(input_df)

    # Convert the predicted encoded label back into the original disease name
    predicted_disease = label_encoder.inverse_transform(predicted_encoded_label)

    return predicted_disease[0]

print("Prediction function `predict_disease` defined (with feature names handling).")
```
